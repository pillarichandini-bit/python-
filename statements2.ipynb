```python
# Program to calculate remaining balance after withdrawal

# Initial balance
balance = float(input("Enter your account balance: "))

# Withdrawal amount
withdraw = float(input("Enter withdrawal amount: "))

# Check if withdrawal is possible
if withdraw <= balance:
    remaining_balance = balance - withdraw
    print("Initial Balance:", balance)
    print("Withdrawal Amount:", withdraw)
    print("Remaining Balance:", remaining_balance)
else:
    print("Insufficient funds! Withdrawal not possible.")
```

```python
# Program to calculate marks and attendance

# Input marks
marks_obtained = float(input("Enter marks obtained: "))
total_marks = float(input("Enter total marks: "))

# Input attendance
classes_attended = int(input("Enter number of classes attended: "))
total_classes = int(input("Enter total number of classes: "))

# Calculate percentages
marks_percentage = (marks_obtained / total_marks) * 100
attendance_percentage = (classes_attended / total_classes) * 100

# Display results
print("\n--- Student Report ---")
print("Marks Obtained:", marks_obtained, "/", total_marks)
print("Marks Percentage:", round(marks_percentage, 2), "%")
print("Classes Attended:", classes_attended, "/", total_classes)
print("Attendance Percentage:", round(attendance_percentage, 2), "%")

# Simple condition check
if marks_percentage >= 75 and attendance_percentage >= 75:
    print("Result: PASS ")
else:
    print("Result: FAIL ")

    
```
