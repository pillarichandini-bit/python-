```python
#fail or pass
def find_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "c"
    elif marks >= 60:
        return "D"
    elif marks >= 40:
        return "E"
    else:
        return "fail"

    
marks = int(input("Enter marks: "))
print("Grade:", find_grade(marks))

    
```

```python
#deposit , withdraw
def deposit(balance, amount):
    return balance + amount

def withdraw(balance, amount):
    if amount <= balance:
        return balance - amount
    else:
        return "Insufficient funds"

balance = 1000
print("Initial Balance:", balance)

balance = deposit(balance, 500)   # 1500
print("After Deposit:", balance)

balance = withdraw(balance, 200)  # 1300
print("After Withdrawal:", balance)
```

```python
#largest of three numbers
def largest(a, b, c):
    if a >=b and a >=c:
        return a 
    elif b >= a and b >= c:
        return b
    else:
        return c

a = int(input("enetr a:"))
b = int(input("enetr b:"))
c = int(input("enter c:"))

print("largest =", largest(a, b, c))
```
